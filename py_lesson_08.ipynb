### ライブラリとは？

Pythonにおいての**ライブラリ**とは、機能を拡張することができるツール\
（料理においての調理器具のようなポジション）

Pythonのライブラリは多種多様（Pythonが人気である理由の一つでもある）\
例：\
・**numpy**：行列の演算操作\
・**Pandas**：データフレームの読み込み\
・**Tensorflow, Pytorch**：深層学習モデルの実装

使いたいライブラリを自分のPC取り入れる方法も簡単\

### Numpyでの二次元配列

Numpyで二次元配列を作る場合：\
（１）：**x = np.array(range(N))**で、N個の一次元配列を作成\
（２）：**x.reshape(a, b)**で、a行b列の二次元配列に変換

下のお手本に従って実装してみよう

生成した二次元配列xにて\
配列の行と列の数を知りたい場合は以下の２通りがある：\
（１）：**len**を使って取得する：\
・len(x)：行数取得
・len(x[適当な添え字])：列数取得\
（２）：**x.shape**で、一挙に取得する

下のお手本に従って実装してみよう

二次元配列でのアクセスに関して

**x[a, b]（あるいは x[a][b]）**：\
・xのa行b列目の要素にアクセスする\
・添え字は、正の添え字 と 負の添え字でも行える

**x[a]**：\
・xのa行目の要素を一挙にアクセスする

下のお手本に従って実装してみよう

二次元配列xでのスライスに関して

**x[a:b]**：\
・配列a行〜b-1行目までの行列を取り出す

**x[:,c:d]**：\
・配列c列〜d-1列目までの行列を取り出す

**x[a:b,c:d]**：\
・配列a行〜b-1行目、c列〜d-1列目にあたる行列を取り出す

下のお手本に従って実装してみよう

arrayとreshapeをつなげると、二次元配列が次の１文でも行うことができる

**x = np.array(range(a, b)).reshape(p, q)**：\
・a~b-1の要素を持つ、p行q列の二次元配列xを生成\
ただし、配列xの要素数 == p * q でないとエラーが出る

下のお手本に従って実装してみよう

二次元配列で乱数を作る場合

（１）整数で作る場合：

**np.random.randint(x, y, N).reshape(a, b)**：\
・前半でも出てきた一次元配列での乱数生成に、reshapeを加えることa行b列の整数の二次元配列ができる

（２）実数で作る場合：

**np.random.rand(a, b)**：\
・0以上１未満の数値でa行b列の実数の二次元配列を作る

**np.random.rand(a, b) * (max - min) + min**：\
・min以上max未満の数値でa行b列の実数の二次元配列を作る

下のお手本に従って実装してみよう

配列に関しての総和は、実は関数を使えば簡単に求めることができる

配列x全体の総和を求める場合：

**np.sum(x)**：\
・配列xにある全要素の総和を出す

二次元配列xの場合：

**np.sum(x, axis=0)**：\
・各**列**での総和を求める

**np.sum(x, axis=1)**：\
・各**行**での総和を求める
