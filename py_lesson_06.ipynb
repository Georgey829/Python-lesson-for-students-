### while文による繰り返し

while文の役割：\
・特定の条件が満たしていない限り、処理を繰り返し行わせることができる\
（特に、どのくらいの回数で終えるべきかを決めていない、あるいは想定できないなどには便利）\
・ソフトウェアでも、ゲームでも専ら用いられる（日常生活にもある）

例：\
・RPGにおいての、名前の入力待ち\
（条件：入力が終わるまで先が進まない）\
・どちらかが2回勝つまでの３本勝負\
（条件：どちらかが連続で2回勝って終わる場合や、引き分けが永遠に続くこともある）\

**while文**の書き方：\
・for文では、**（）内に条件を設けた条件式を使って繰り返し処理を行う**\
・**条件式が[True]である限り繰り返す**\
・if文、for文と同じく、**while文**の後に繰り返す処理の内容はインデントしてから書く
