### 変数の型の種類

*Pythonについている基本的な型は主に４つ*

**int：整数（eg: 1, -1, 0, 123, ...）**\
**float：浮動小数点（eg: 1.1, -2.1, 0.0, 1/2, ...）**\
**string：文字列（eg: "UTokyo", "東京大学", ...）**\
**bool：論理値（eg: True, False）**

変数が何の型なのか確かめるには、関数*type*を使う

### 変数の型変換

前回にて、**文字列と数値といった異なる型の組み合わせでは、演算子を使った計算はできない**ことを紹介

型が同じ同士のものは演算子を使って計算が可能 **（型変換：type casting）**

**valueが文字型のままで、計算を行うことによるエラーが出てくる**

**int型への型変換は変数に対して *int()* で囲う**

ただし、数字ではない*str型*の変数を変換しようとするエラーが出る

**float型への型変換は変数に対して *float()* で囲う**

ただし、数字ではない*str型*の変数を変換しようとするエラーが出る

**str型への型変換は変数に対して *str()* で囲う**

**bool型への型変換は変数に対して *bool()* で囲う**
