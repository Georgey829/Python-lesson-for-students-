### 条件分岐と比較演算子

条件分岐・繰り返し処理では、その処理を行うための式が成立しているかをチェックした上で行う

**比較演算子を使いこなすのが重要となっていく！**

例：\
'=='：等しい\
'!='：等しくない\
'>'：より大きい\
'<'：より小さい\
'>='：以上\
'<='：以下

では、早速、比較演算子を使っての結果を見てみよう

Pythonでの比較演算子の置き方として：\
・変数と変数の間に設ける\
・変数と数値の間に設ける\
・数式と数値の間に設ける

下のお手本に従って実装してみよう

```python
a = 1
b = 2
c = a

print(a == b)
print(a == c)
print(a >= c)
print(a > c)
```

### if文を使った条件分岐（if-else編）

if文の役割：\
・条件に応じて、別々の処理を行うことができる\
・TrueかFalseかで処理内容を変えることができる\
・ソフトウェアでも、ゲームでも用いられる

Pythonでの*if文の書き方*

if文による条件分岐では、*if()*の括弧内に条件式を書いた後、':'を書く\
ただし、if文内で行ってほしい処理はインデント（字下げ）で示す

下のお手本に従ってコードを書いてみましょう

```python
x = 10
y = 10

print("Start")
if(x == y):
    print("x is equal to y")
print("End")
```

Pythonでのif文の書き方 **（elseも使う場合）**

条件分岐に沿わない場合、別の処理を行わせたい場合は*else*を使う\
*else*では、*else*の後に、':'を書く\
*else文*内で行ってほしい処理はインデント（字下げ）で示す

下のお手本に従ってコードを書いてみよう

```python
x = 10
y = 3

print("Start")
if(x == y):
    print("x is equal to y")
else:
    print("x is not equal to y")
print("End")
```

今度は、*if文*の条件式に数式を設けてみる\
ただし、数式に用いられる演算子に関しては2回目を参照！

下のお手本に従って実装してみよう

```python
value = input("Input a number: ")
value = int(value)

if(value % 2 == 0):
    print("The number is even")
else:
    print("The number is odd")
```

*if文*のさらに内側に*if文*をさらに設けることができる\
ただし、さらに*if文*を設ける場合は、インデントがまた必要になる

下のお手本に従って実装してみよう

```python
vale = input("Input a number: ")
value = int(value)

if(value % 2 == 0):
    print("The number is even")
    
    if(value % 3 == 0):
        print("The number is also divisible by 3")
else:
    print("The number is odd")
```

### if文を使った条件分岐（if-elif-else編）

条件によっては、最初の条件は満たしていなくても、別の条件を満たしていれば、行ってほしい場合がある

Pythonでの*elif文の書き方*

*elif文*による条件分岐では、'elif()'の括弧内に条件式を書いた後、':'を書く\
*elif文*内で行ってほしい処理もインデント（字下げ）で示す

**注意：elifは、if文の処理の下に書くのが基本**\
ただし、if文がないままelifを始める、あるいはelseの後にelifを書くとエラーが出る

下のお手本に従ってコードを書いてみよう

```python
x = 5
y = 10
z = 3

if(x > y):
    print("x > y")
elif(x >= z):
    print("y >= x >= z")
else:
    print("x < z")
```

**elifの処理の中（あるいはelseの処理の中）にさらに、if文を設けることは可能！**

その分、適用させるべきインデントも増えるので注意してコーディングをしよう

下のお手本に従って実装してみよう

```python
value = input("Input a number: ")
value = int(value)

if(value % 2 == 0):
    print("The number is even")

    if(value == 0):
        print("Then the number is zero")
    elif(value > 0):
        print("Then the number is greater than 0")
    else:
        print("Then the number is less than 0")

else:
    print("The number is odd")

    if(value == 0):
        print("Then the number is zero")
    elif(value > 0):
        print("Then the number is greater than 0")
    else:
        print("Then the number is less than 0")
```

### if-else と if-elif-else　の使い分け

条件分岐において、if-elseだけでは「仕様内容」によってうまく動作しないことがある

下のお手本に従ってコーディングしてみよう\
**特に score=90 とした時の結果を見てみよう**

```python
score = input("Input a score: ")
score = int(score)

if(score >= 90):
    grade = 'A'
if(score >= 80):
    grade = 'B'
else:
    grade = 'C'

print("Your grade is {}".format(grade))
```

90点以上のscoreを入れてもgradeがBになる？

なぜ？\
if文のみの実装では、順次条件分岐を参照するから......（ここで説明）

今度は、実装内容に沿ってのコーディングを行わせる

下のお手本に従って実装してみよう

```python
score = input("Input a score: ")
score = int(score)

if(score >= 90):
    grade = 'A'
elif(score >= 80):
    grade = 'B'
else:
    grade = 'C'

print("Your grade is {}".format(grade))
```
