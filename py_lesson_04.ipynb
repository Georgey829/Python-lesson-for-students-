・*elif文*による条件分岐では、*elif()*の括弧内に条件式を書いた後、':'を書く\

**elif文は、if文の処理の下に書くのが基本！**

### 複数の条件を設ける（AND）

条件分岐においては、「ある条件」と「ある条件」が満たしている場合に実施してほしい処理がある

例：\
・「年齢が18歳以上」で「学生」の場合 → 学割を適用\
・「会計が10,000円以上」で「予約人数が20人以上」 → 団体割引を適用\
**すなわち、条件らすべてが「True」となる場合**

**and**:\
・「**式Aかつ式B**が真となる場合」を適用するときに使う\
・式のうち１つでも偽である場合、その条件式は偽となる

Pythonの条件分岐において、*and*を適用させる場合は：

**if(条件式１)and(条件式２)**

注意点：\
・条件式の間に*and*を設ける\
・条件の末尾に':'を忘れないように

### 複数の条件を設ける（OR）

条件分岐において、「ある条件」あるいは「ある条件」のどちらかが満たしている場合に実施してして欲しい処理がある

例：\
・「会計が30,000円以上」あるいは「買い物回数が年３回以上」の場合 →  VIP待遇を適用\
・「全試合で３勝以上」あるいは「ゴールした総数が10回以上」 → ランク昇格を適用\
**すなわち、どちらかの条件が「True」となる場合**

**or**:\
・「式Aか式Bが真となる場合（一方が真である場合）」を適用する時に使う\
・式のうち１つでも真がある場合、その条件式は真になる

Pythonの条件分岐において、*or*を適用させる場合は：

**if(条件式１)or(条件式２)**

注意点：\
・条件式の間に*or*を設ける\
・条件の末尾に':'を忘れないように
