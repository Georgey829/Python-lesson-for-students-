### for文の繰り返し

for文の役割:\
・繰り返す回数を設けての、処理を連続して行うことができる\
（処理が同じなら、同じ命令文を書くより繰り返しを使った方が楽）\
・ソフトウェアでも、ゲームでも専ら用いられる\

例：\
・（形式が決まっている）顧客情報の入力\
・フレーズの繰り返し

**for文の書き方：**\
・for文では、回数を数える役割を持たせた変数といつまで繰り返すかを決める変数を使って繰り返し処理を行う

**for i in range(num)**\
・iがループを数える変数\
・range(num)が何回繰り返すのかを決める変数

**要注意：条件分岐と同じく、for文の後に繰り返す処理の内容はインデントしてから書く**

（２）いわゆる省略された書き方になっている

for文に*if文*を入れることで、条件に応じた処理を繰り返し行わせることができる\
（条件に応じて、別々の処理を行うことができる）

ソフトウェアでも、ゲームでも専ら用いられる（日常生活にもある）：\

### for文による繰り返し（途中で抜けるあるいは処理を飛ばす）

for文で実施している最中、エラーが生じてしまうとその処理自体が止まってしまう...

とりあえず、下のお手本に注目：

上記のエラーの内容は、i=0の状態で割り算を出力させようとしてのエラーになっているので\
分母が0がになってもエラーを出さないためには...

・**（方法一）割り算をやめさせる**\
・**（方法二）0による割り算の実施という処理を飛ばす**

繰り返し処理の中で、途中で処理を終わらせる場合は**break**を使う\
（大抵はif文と組み合わせて用いられる）

**break**が適用されると、繰り返し処理の途中にもかかわらず、for文から抜ける\
要注意：**break**以降の、繰り返し処理内の内容が行われない！

（大抵はif文と組み合わせて用いられる）

**continue**が適用されると、以降の処理は飛ばされ、カウンターが更新\
再度繰り返し処理が実施される\
（ただし、**break**とは異なり、繰り返し処理自体を抜けることはないところに注目）
