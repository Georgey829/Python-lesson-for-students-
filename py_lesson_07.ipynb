### Pythonの標準ライブラリ

Pythonで予め埋め込まれているツールを*標準ライブラリ*と呼ぶ\
・ただし、ライブラリは特殊な機能を持った関数やクラスを一つにまとめたもの、機能を拡張できるツール\
（ちなみに、最近ホットな、人工知能実装のためのライブラリもある！）

ライブラリ内には、実行可能なコードファイルである*モジュール*が格納されている

今回、\
・乱数を生成できるライブラリ\
・数式を呼び出せるライブラリ

**標準ライブラリの呼び出し方は：import ライブラリ名**

**標準ライブラリ内の関数を呼び出す場合は：ライブラリ名.関数名()**
